```python
import pandas as pd
import numpy as np
from statsmodels.tsa.regime_switching.markov_regression import MarkovRegression
import matplotlib.pyplot as plt
df = pd.read_excel(r'/content/drive/MyDrive/Trabalho Final Setor Publico/Mexico_Policy_Uncertainty_Data.xlsx')
# Garantir tipos corretos
df['Year'] = df['Year'].astype(int)
df['Month'] = df['Month'].astype(int)

# Criar coluna de data
df['Date'] = pd.to_datetime(dict(year=df.Year, month=df.Month, day=1))
df = df.set_index('Date')

df = df[df.index >= '1990-07-01']

# Série mensal de EPU
epu = df['Mexican Policy Uncertainty Index'].dropna()

# Normalizar a série
epu_norm = (epu - epu.mean()) / epu.std()

# Estimar modelo Markov Switching com variância heterocedástica
model = MarkovRegression(epu_norm, k_regimes=2, trend='c', switching_variance=True)
result = model.fit(method='powell', maxiter=1000)

# Obter probabilidades suavizadas do regime 2 (alta incerteza)
smoothed_probs = result.smoothed_marginal_probabilities[1]

# Plotar os resultados
plt.figure(figsize=(12, 6))
plt.plot(epu_norm, label='EPU (normalizado)')
plt.plot(smoothed_probs, label='Prob. regime 2 (alta incerteza)', color='red', alpha=0.7)
plt.title('Markov Switching no EPU - México (mensal)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
